```python
# ============imports============
import matplotlib.pyplot as plt
import pandas as pd
from data_provider import ROOT
from pathlib import Path

csv_report_path = ROOT / "reports" / "csv"
```

# base-CNN

```python
plot_path = csv_report_path / "base_model" / "CNN-BaseModel.csv"

report = pd.read_csv(plot_path)

plt.plot(range(1, 9), report["train_loss"], label="Train Loss", c="green", marker="o")
plt.plot(range(1, 9), report["val_loss"], label="Validation Loss", c="red", marker="o")
plt.legend()
plt.grid(True)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.show()
```
